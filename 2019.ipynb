```python
urls = {
    2019: "https://datagov.mot.go.th/dataset/7e077ffd-dc4f-4dc6-a71c-0813726f3c12/resource/4d919e5b-aa06-4fe4-8a13-5571747a1254/download/accident2019.csv",
    2020: "https://datagov.mot.go.th/dataset/7e077ffd-dc4f-4dc6-a71c-0813726f3c12/resource/40476c7b-4194-4dc2-aba5-7326819ed071/download/accident2020.csv",
    2021: "https://datagov.mot.go.th/dataset/7e077ffd-dc4f-4dc6-a71c-0813726f3c12/resource/d0a8601a-3ee5-4ce8-8e9d-3dae3915dc30/download/accident2021.csv",
    2022: "https://datagov.mot.go.th/dataset/7e077ffd-dc4f-4dc6-a71c-0813726f3c12/resource/733b7874-bd5f-44b9-b271-650890b061f2/download/accident2022.csv",
    2023: "https://datagov.mot.go.th/dataset/7e077ffd-dc4f-4dc6-a71c-0813726f3c12/resource/661f2ead-1f28-4cd9-8a67-1bef76b33ef6/download/accident2023.csv",
    2024: "https://datagov.mot.go.th/dataset/7e077ffd-dc4f-4dc6-a71c-0813726f3c12/resource/83e9fae0-5f33-45e7-9a6d-6d5ad2060a08/download/accident2024.csv",
    2025: "https://datagov.mot.go.th/dataset/7e077ffd-dc4f-4dc6-a71c-0813726f3c12/resource/e1db5e93-2b70-4ee4-b6d7-7ead76d14a09/download/accident2025.csv",
    2026: "https://datagov.mot.go.th/dataset/7e077ffd-dc4f-4dc6-a71c-0813726f3c12/resource/4625b7aa-99a4-4dfb-9a0f-6402814d0f2c/download/accident2026.csv"
}
```

```python
def clean_accident_data(df):

    df = df.dropna(how="all").copy()


    columns_to_drop = [
        "วันที่รายงาน",
        "เวลาที่รายงาน",
        "หน่วยงาน",
        "บริเวณที่เกิดเหตุ/ลักษณะทาง",
        "มูลเหตุสันนิษฐาน",
        "ลักษณะการเกิดอุบัติเหตุ",
        "สภาพอากาศ",
        "รถคันที่ 1",
        "ACC_CODE",
        "สายทางหน่วยงาน",
        "รหัสสายทาง",
        "รถคันที่1",
        "บริเวณที่เกิดเหตุ",
        "ลักษณะการเกิดเหตุ",
        "รถบรรทุก6ล้อ",
        "รถบรรทุกไม่เกิน10ล้อ",
        "รถบรรทุกมากกว่า10ล้อ",
        "รถอีแต๋น",
        "รถอื่นๆ",
        "คนเดินเท้า",
        "ผู้บาดเจ็บเล็กน้อย",
        "ผู้บาดเจ็บสาหัส",
        "รถจักรยานยนต์",
        "รถสามล้อเครื่อง",
        "รถยนต์นั่งส่วนบุคคล",
        "รถตู้",
        "รถปิคอัพโดยสาร",
        "รถโดยสารมากกว่า4ล้อ",
        "รถปิคอัพบรรทุก4ล้อ",
        "รถและคนที่เกิดเหตุ",
        "จำนวนที่เกิดเหตุทั้งหมด(รวมคนเดินเท้า)",
        "รถยนต์นั่งส่วนบุคคล/รถยนต์นั่งสาธารณะ",
        "รถบรรทุกมากกว่า6ล้อไม่เกิน10ล้อ",
        "รถบรรทุกมากกว่า10ล้อ(รถพ่วง)",
        "อื่นๆ",
        "จำนวนผู้บาดเจ็บสาหัส",
        "จำนวนผู้บาดเจ็บเล็กน้อย",
        "บริเวณที่เกิดเหตุอื่น"
]

    df = df.drop(
        columns=columns_to_drop,
        errors="ignore"
    )

    df = df.dropna(
        subset=["LATITUDE", "LONGITUDE"]
    ).copy()

    return df
```

```python
cleaned_data = {}

for year, url in urls.items():
    print(f"กำลัง Clean ข้อมูลปี {year}...")


    df = pd.read_csv(url)


    df_clean = clean_accident_data(df)


    cleaned_data[year] = df_clean

    print(f"ปี {year}: เหลือ {len(df_clean):,} แถว")

print("Clean ข้อมูลครบทุกปีแล้ว")
```
