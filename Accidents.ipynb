```python
df = df.dropna(how="all")
print("จำนวนแถวหลังลบ:", len(df))
display(df.tail())
```

```python
missing_count = df.isnull().sum()
display(missing_count[missing_count > 0].sort_values(ascending=False))
```

```python
rows_with_missing = df[df.isnull().any(axis=1)]
display(rows_with_missing.head(10))
```

```python
df_clean = df.copy()
columns_to_drop = [
    "วันที่รายงาน",
    "เวลาที่รายงาน",
    "หน่วยงาน",
    "บริเวณที่เกิดเหตุ/ลักษณะทาง",
    "มูลเหตุสันนิษฐาน",
    "ลักษณะการเกิดอุบัติเหตุ",
    "สภาพอากาศ",
    "รถคันที่ 1"
    ]

df_clean = df_clean.drop(columns=columns_to_drop)
print("จำนวนแถวและคอลัมน์หลังลบ:", df_clean.shape)
print("คอลัมน์ที่เหลือ:")
print(df_clean.columns.tolist())
```

```python
missing_count = df_clean.isnull().sum()
display(missing_count[missing_count > 0].sort_values(ascending=False))
```

```python
rows_with_missing_clean = df_clean[df.isnull().any(axis=1)]
display(rows_with_missing_clean.head(10))
```

```python
missing_latitude = df_clean[df_clean["LATITUDE"].isnull()]
display(missing_latitude)
```

```python
df_clean = df_clean.dropna(subset=["LATITUDE"])
missing_count = df_clean.isnull().sum()
display(missing_count[missing_count > 0].sort_values(ascending=False))
```
